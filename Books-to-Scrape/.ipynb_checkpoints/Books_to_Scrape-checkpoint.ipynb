```python
import requests
from bs4 import BeautifulSoup

base_url = "http://books.toscrape.com/catalogue/page-{}.html"

for page in range(1, 10):
    url = base_url.format(page)

    response = requests.get(url)
    soup = BeautifulSoup(response.text, "lxml")

    for rating in soup.select(".star-rating"):
        print(rating["class"][1])
```

```python
website = soup.select(".product_pod")
```

```python
book_name = []
book_price = []
book_rating = []
is_book_available = []

for i in website:
    book_name.append(i.find("h3").find("a")["title"])
    book_price.append(i.find("p", class_="price_color").text.strip())
    book_rating.append(i.find("p", class_="star-rating")["class"][1])
    is_book_available.append(i.find("p", class_ = "instock availability").text.strip())

data = {"book_name":book_name, "book_price":book_price, "book_rating":book_rating, "is_book_available":is_book_available}
df = pd.DataFrame(data)
df
```

```python
book_data_set = pd.DataFrame()

for j in range(1, 30):
    url = f"http://books.toscrape.com/catalogue/page-{j}.html"
    header = {
    "User-Agent" : "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/145.0.0.0 Safari/537.36"
    }
    book_ = requests.get(url, headers = header).text
    soup = BeautifulSoup(book_, "lxml")
    website = soup.select(".product_pod")

    book_name = []
    book_price = []
    book_rating = []
    is_book_available = []
    
    for i in website:
        book_name.append(i.find("h3").find("a")["title"])
        book_price.append(i.find("p", class_="price_color").text.strip())
        book_rating.append(i.find("p", class_="star-rating")["class"][1])
        is_book_available.append(i.find("p", class_ = "instock availability").text.strip())

    data = {"book_name":book_name, "book_price":book_price, "book_rating":book_rating, "is_book_available":is_book_available}
    df = pd.DataFrame(data)
    final = pd.concat([book_data_set, df], ignore_index=True)
print(final)
```

```python
final.to_csv("books_data_set.csv", index=False)
```
