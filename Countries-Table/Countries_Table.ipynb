```python
Countries_of_the_World = pd.DataFrame()


url = "https://www.scrapethissite.com/pages/simple/"
header = {
    "User-Agent" : "user-agentMozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/146.0.0.0 Safari/537.36"
}

website_page = requests.get(url, headers = header).text
soup = BeautifulSoup(website_page, "lxml")

    
#name of the countries 
countries_name = []
#capital of the countries
countries_capital = []
#population of the countries
countries_popultaion = []
#area of the countries
countries_area = []

countries = soup.find_all("div", class_="col-md-4 country")

for j in countries:
    countries_name.append(j.find("h3", class_="country-name").text.strip())
    countries_capital.append(j.find("span", class_ = "country-capital").text.strip())
    countries_popultaion.append(j.find("span", class_ = "country-population").text.strip())
    countries_area.append(j.find( "span", class_ = "country-area").text.strip())

data = {
    "countries_name":countries_name, "countries_capital":countries_capital, "countries_popultaion":countries_popultaion, "countries_area":countries_area
}
df = pd.DataFrame(data)
final = pd.concat([Countries_of_the_World, df], ignore_index = True)
print(final)
final.to_csv("countries.csv", index=False)
final.to_json("countries.json", index=False)
```
